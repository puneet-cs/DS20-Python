# match case

```python
# Match case is used to replace multiple if else condition
# Match case is more powerful than the Switch case statment
# Match case works with variable and data structurs. 
```

```python
# Example-1:
num = int(input("Enter a number"))

match num:
    case 10:
        print("You entered 10")
    case 20:
        print("You entered 20")
    case 30:
        print("You entered 30")
    case _:
        print("something else")
```

```python
Biryani = input("Enter Biryani name")

match Biryani:
    case "Paradise":
        print("Spicy & Tasty")
    case "Shahdab":
        print("Double masala")
    case "road side":
        print("Budget friendly")
    case _:
        print("PG Biryani")
```

```python
# Example-2: passing data structure to match case

data = [1,2,3,4]

match data:
    case [4,3,2,1]:
        print("Hello")
    case [1,2,3,4]:
        print("world")
```

```python
# Example-3: if condition with match case

val = int(input("Enter a number"))

match val:
    case x if x > 0:
        print("It is a positive numner")
    case y if y < 0:
        print("It is a Negative Number")
    case _:
        print("It is a Zero")
```

# Insert, delete and update operations indirectly

```python
tuple1 = (10, 20.5, "Hello", [10, 20, 30])
```

```python
tuple1[3][2] = 5000
```

```python
tuple1
```

```python
# tuple items cannot be updated directly
# complete item in tuple cannot be updated directly
# but if item is mutable then elements inside this mutable item can be updated
```

```python
#insert
tuple3 = (10,20,30)
list1 = list(tuple3)
print(list1)
list1.extend([100, 200, 300])
print(list1)
tuple3 = tuple(list1)
print(tuple3)
```

```python
#delete
tuple3 = (10,20,30)
list1 = list(tuple3)
print(list1)
list1.remove(30)
print(list1)
tuple3 = tuple(list1)
print(tuple3)
```

```python
# update
tuple3 = (10,20,30)
list1 = list(tuple3)
print(list1)
list1[2] = 300000
print(list1)
tuple3 = tuple(list1)
print(tuple3)
```

# popular inbuilt function for tuple

```python
Salary = (6, 70, 100, 600, 700, 90, 10, 80, 120)

print(tuple(reversed(Salary)))
print(len(Salary))
print(min(Salary))
print(max(Salary))
print(sum(Salary))

sorted_list = tuple(sorted(Salary, reverse = True))
print(sorted_list)

# sort() is inplace function
# inplace means sorting is done in the same data sturcture

# sorted() function return the list
# sort() function don't return anything
```

```python
l1 = [34, 40, 50, 60, 1, 3, 4, 70]

l1.sort()
print(l1)
```
