# Type casting

```python
# integer typecasting : int()
v1 = int(5.6)  # float to int
v2 = int("500") # string to int
v3 = int(True) # Bool to Int
v4 = int(False) # Bool to Int
v5 = int("1010", 2)

print(v1, type(v1))
print(v2, type(v2))
print(v3, type(v3))
print(v4, type(v4))
print(v5, type(v5))
```

```python
# Convert to float: float()
v1 = float(105)            # int to float
v2 = float("4.5")
v3 = float(True)

print(v1, type(v1))
print(v2, type(v2))
print(v3, type(v3))
```

```python
# convert to string: str()
v1 = str(200)
v2 = str(4.5)
v3 = str(True)

print(v1, type(v1))
print(v2, type(v2))
print(v3, type(v3))
```

```python
# Convert to Boolean: bool()
v1 = bool(1)
v2 = bool(0)
v3 = bool(1000)
v4 = bool("Hello")

print(v1, type(v1))
print(v2, type(v2))
print(v3, type(v3))
print(v4, type(v4))
```

# 2. Bitwise operators (<<, >>, ~, &, |, ^)

```python
# Bitwise operators works on binary number
# Bitwise operators are faster than arithmatic operators
# Bitwise operators perform many operation in easy way due to binary number system.
```

```python
a = 10
b = 3

print(a & b)
print(a | b)
print(a ^ b)
```

```python
a = 10
b = 3
c = 2
print(a & b & c)
```

```python
print(a | b | c)
```

```python
print(a ^ b ^ c)
```

#### left shift (<<) , Right shift (>>)

```python
# Def (left shift): Insert new bits (0) to the LSB
# Def (Right shift) : Delete the bits from the LSB
```

```python
a = 10
print(a << 1)  # left shift by 1 bit
```

```python
print(a << 2)  # left shift by 2 bit
```

```python
print(a << 3)  # left shift by 3 bit
```

```python
print(a << 6)  # left shift by 8 bit
```

```python
# Right shift (>>)

a = 10
print(a >> 1)
```

```python
print(a >> 2)
```

```python
print(a >> 3)
```
