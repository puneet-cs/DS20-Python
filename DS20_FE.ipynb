#  Feature Engineering

```python
# Def: Feature engineering helps to improve the performance of ML models.

# ML models works as mathematical equations.
# Mathematical equations cannot be apply on categorical data.
# In FE we convert categorical data to Numerical data

# Feature scaling : to bring the data in a particular range
```

### Feature Encoding

```python
# converting categorical data to numerical data

# Populor method for feature encoding:
# 1. Label encoding : when their is no order between the categories
# 2. Ordinal encoding : when their is order between the categories
# 3. One hot encoding  :use when number of entires is less in categorical column
```

# 1. Label encoding 

```python
# Def: Assign an integer to each category in Lexographical order
!pip install scikit-learn
```

```python
# Example-1

from sklearn.preprocessing import LabelEncoder
import pandas as pd

data = {'Color' : ['Telupu' ,'Arepu', 'Paspu', 'Neelum', "Akupacha"]}

df = pd.DataFrame(data)

# create the labelenocder object
encoder = LabelEncoder()

# fit() function allot the integer value to each category
encoder.fit(df['Color'])

# transform() function transform the actual data based upon fit() function allocation
df['encoded_color'] = encoder.transform(df['Color'])

print(df)
```

# 2. Ordinal encoding

```python
# Convert categorical data into numerical data based upon the ranking.

from sklearn.preprocessing import OrdinalEncoder

data = [['Phd'], ['UG'], ['MS'], ['UG'], ['Phd']]

encoder = OrdinalEncoder(categories = [['Phd', 'UG', 'MS']])

encoded_data = encoder.fit_transform(data)
print(encoded_data)
```

```python
# Use label encoder when we don't have ranking in the categories.
# Use Ordinal encoder when we have ranking in the categories.
```

# One hot encoder

```python
# convert categorical data into binary indicators.

data = {"country" : ['India', 'USA', 'Uganda', 'Japan', 'Thailand', 'Nigeria', 'India']}

df = pd.DataFrame(data)

encoded_data = pd.get_dummies(df, dtype = int)

encoded_data
```

```python
df
```

```python
# Machine LOVE the one hot encoder due to binary encoding
# Don't use one hot encoder, when number of categories are large
# Use one hot encoder when having less number of categories
```

# Method-2 : one hot encoding using sklearn

```python
from sklearn.preprocessing import OneHotEncoder

encoder = OneHotEncoder(sparse_output = False, dtype = int)

encoded_data = encoder.fit_transform(df[['country']])
print(encoded_data)
```

# Feature scaling : to bring the data in a particular range

```python
# ML models give more importance to large values
# So we bring all the data into same range. 0 to 1

# type of feature scaling:
# 1. Min Max scaling
# 2. Standardization (Z-Score scaling)
# 3. Robust scaling
```

# 1. Min Max scaling

```python
# Min max scaler bring the value by deafult in 0 to 1 range.
# X_scaled = (X - X_min) / (X_max - X_min)

# X = [ 100000, 200000, 300000, 400000, 500000 ] 
# X = 200000 -> 0.25
# X = 300000 => (300000 - 100000) / (500000 - 100000) = 0.5
# X = 400000 -> 0.75
# X = 100000 -> 0
# X = 500000 -> 1
```

```python
from sklearn.preprocessing import MinMaxScaler
import pandas as pd

scaler = MinMaxScaler()

data = [ 100, 200, 300, 400, 500 ] 

df = pd.DataFrame(data, columns = ["values"])
df['scaled_values'] = scaler.fit_transform(df[['values']])
df

# MinMaxSclaer cannot handle the outlier
```

# Standardization ( Z-Score scaler)

```python
# Z-Score scaler is used when data is Normally distributed
# Z_score = (X - mean) / SD

# Z_score scaler will bring the value in the range of -3 to +3

from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

data = [ 100, 200, 300, 400, 500 ] 
df = pd.DataFrame(data, columns = ["Values"])

df["scaled_values"] = scaler.fit_transform(df[["Values"]])
df
```

```python

import numpy as np
mean = np.mean(df.Values)
SD = np.std(df.Values)
X = 100
Z_score = (X - mean) / SD
print(Z_score)
```

# Robust Scaler

```python
# Use the Robust scaler when their is Outliers in the dataset
# Robust scaler is based upon IQR and median

# X_scaled = (X - Median(X))/ IQR(X)

# IQR : Inter Quartile Range = Q3 - Q1

from sklearn.preprocessing import RobustScaler

scaler = RobustScaler()

data = [ 100, 200, 300, 400, 5000 ] 
df = pd.DataFrame(data, columns = ["Values"])

df['scaled_value'] = scaler.fit_transform(df[['Values']])
df
```

```python
# Manually
# X_scaled = (X - Median(X))/ IQR(X)
Median = np.median(df.Values)
Q1 = np.percentile(df.Values, 25)
Q3 = np.percentile(df.Values, 75)
print(Q1, Q3)
IQR = Q3 - Q1
X = 100

X_scaled = (X - Median) / IQR
X_scaled
```
