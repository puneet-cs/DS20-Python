# Type casting

```python
# integer typecasting : int()
v1 = int(5.6)  # float to int
v2 = int("500") # string to int
v3 = int(True) # Bool to Int
v4 = int(False) # Bool to Int
v5 = int("1010", 2)

print(v1, type(v1))
print(v2, type(v2))
print(v3, type(v3))
print(v4, type(v4))
print(v5, type(v5))
```

```python
# Convert to float: float()
v1 = float(105)            # int to float
v2 = float("4.5")
v3 = float(True)

print(v1, type(v1))
print(v2, type(v2))
print(v3, type(v3))
```

```python
# convert to string: str()
v1 = str(200)
v2 = str(4.5)
v3 = str(True)

print(v1, type(v1))
print(v2, type(v2))
print(v3, type(v3))
```

```python
# Convert to Boolean: bool()
v1 = bool(1)
v2 = bool(0)
v3 = bool(1000)
v4 = bool("Hello")

print(v1, type(v1))
print(v2, type(v2))
print(v3, type(v3))
print(v4, type(v4))
```

# Output in Python

```python
# 1. concatenation using commas
# 2. String concatenation using +
# 3. str.format() method
# 4. f-strings method (python 3.6+ version)
# 5. Escape sequences
# 6. End and Sep parametes in print()
```

```python
# 1. concatenation using commas

Hero_name = 'Prabhas'
age = 40
fee = 100

print(Hero_name,age,fee)
```

```python
# 2. String concatenation using +
Hero_name = 'Prabhas'
Heroine_name = 'Anushka'
fee = 100

print(Hero_name + " " + Heroine_name +" " + str(fee))
```

```python
# 3. str.format() method
Hero_name = 'NTR'
age = 40

print("Name: {1}   Age: {0}".format(age, Hero_name))
```

```python
# 4. f-strings method (python 3.6+ version)
Hero_name = 'NTR'
age = 40

print(f"Name of the hero@: {Hero_name} , Age: {age}")
```

```python
# 5. Escape sequence in print()
Hero_name = 'Prabhas'
Heroine_name = 'Anushka'
fee = 100

print(f"Hero Name: {Hero_name}, \t Heroine name: {Heroine_name}, \t FEE : {fee}")
```

```python
print(f"Hero Name: {Hero_name}, \nHeroine name: {Heroine_name}, \nFEE : {fee}")
```

```python
print(f"Hero Name: \"{Hero_name}\", \nHeroine name: {Heroine_name}, \nFEE : {fee}")
```

```python
# 6. End and Sep parametes in print()
Hero_name = 'Prabhas'
Heroine_name = 'Anushka'
fee = 100

print(Hero_name, end = "  ")
print(Heroine_name)
```

```python
print(Hero_name,Heroine_name,fee, sep = " @ ")
```

```python
# end parameter is used to decide how the print statement will END.
# sep parameter decides how the multiple values will be separated
```
