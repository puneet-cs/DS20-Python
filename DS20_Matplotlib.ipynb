# Bar chart

```python
# USE: used to shows the discrete values of categorical data

food_items = ["Idli", "Dosa", "Vada", "Sambhar", "Puri"]
like_count = [2, 7, 3, 10, 12]

plt.figure(figsize = (3,2))

plt.bar(food_items, like_count, color = ["orange", "blue", "purple", "grey", "black"])

for i, count in enumerate(like_count):
    plt.text(i, count+0.2, count, ha = "center")  # plt.text(x_index, y_index, value, position)

plt.xlabel("food items")
plt.ylabel("Number of likes")
```

# Histogram

```python
# Histogram works on continues data
# Histogram is popularly used to find the shape of the data

import numpy as np

np.random.seed(42)
data = np.random.randn(1000)  # normal distributed data : range = -3 to +3 => 99.7%

plt.figure(figsize = (6,4))

plt.hist(data, edgecolor = 'black', bins = 5)
plt.title("Histogram chart")
plt.xlabel("X-data")
plt.ylabel("y-data")
plt.show()
```

```python
# Interview Question: What is the difference between Bar chart and Histogram?
# Ans: Bar chart works on categorical data while histogram works on continuous data
# Bins in histogram don't have space between them.
```

# Scatter plot

```python
# USE: To find the relationship between two variables
# e.g finding the relationship between age and salary

# type of relationships:
# 1. Postive relation (both variable values increasing or both var value decreasing
# 2. Negative relation (one increase other decrease)
# 3. No relation 

age = [5, 8, 10, 15, 20, 25, 30, 40, 50, 68, 99]
salary = [2, 5, 10, 20, 50, 100, 200, 500, 1000, 2000, 3999]
salary.reverse()
print(salary)

plt.scatter(age, salary, s = 100, color = "black")

plt.title("Age VS Salary")
plt.xlabel("Age")
plt.ylabel("Salary")
```

# Method-1 : using plt.subplot() : good for less number of plots

```python
# plt.subplot(nrows, ncols, index)
from sklearn.datasets import load_digits
import matplotlib.pyplot as plt
digits = load_digits()


mobile_company = ["Realme", "Samsung", "Nokia", "Lava", "Apple", "IQoo", "OnePlus"]
market_share = [15, 25, 2, 3, 30, 5, 20]

food_items = ["Idli", "Dosa", "Vada", "Sambhar", "Puri"]
like_count = [2, 7, 3, 10, 12]

plt.figure(figsize = (6,4))

plt.subplot(2,3,1)
plt.pie(market_share, labels = mobile_company)

plt.subplot(2,3,2)
plt.barh(food_items, like_count, color = ["orange", "blue", "purple", "grey", "black"])

plt.subplot(2,3,3)
plt.imshow(digits.images[4])

plt.subplot(2,3,4)
plt.barh(food_items, like_count, color = ["orange", "blue", "purple", "grey", "black"])

plt.subplot(2,3,6)
plt.barh(food_items, like_count, color = ["orange", "blue", "purple", "grey", "black"])

plt.tight_layout()  # use this if subplots are overlapping
plt.show()
```

# method-2: using plt.subplots() : good for large number of subplots

```python
Dinner_dream_thali = ["Biryani", "Chicken", "Rusgulla", "Roti", "Lassi"]
percentage = [30, 20, 10, 20, 20]

food_items = ["Idli", "Dosa", "Vada", "Sambhar", "Puri"]
like_count = [2, 7, 3, 10, 12]

_, axes = plt.subplots(2,3, figsize= (10,10))  # (nrows, ncols, full image size
titles = ["Gopi", "Hari", "Ajay", "Dinesh", "Asma", "Pavan"]

for title, ax in zip(titles, axes.ravel()):
    if title == "Gopi":
        ax.bar(food_items, like_count)
        ax.set_title(title)
    else:
        ax.pie(percentage, labels = Dinner_dream_thali, autopct = '%1.1f%%')
        ax.set_title(title)
# plt.tight_layout()
plt.show()
```

# 3D charts in matplotlib

```python
import numpy as np

z = np.linspace(0, 10, 100)
x = np.sin(z)
y = np.cos(z)

fig = plt.figure()
ax = fig.add_subplot(111, projection = '3d')
ax.plot(x,y,z)  # line chart
```

# Pairplot 

```python
# USE: Pair plot shows the relationship between all the numerical variable together pair-wise
# Hue parameter in pairplot, is used for categorize the data

sns.pairplot(df, hue = 'time')
```

# Heatmap

```python
# Heatmap is used to visulaize the correlation between 2 variables in colorfull chart.
# df.corr() works on numerical data
corr_matrix = df.corr(numeric_only = True)
print(corr_matrix)
```

```python
plt.figure(figsize = (3,2))
sns.heatmap(corr_matrix, annot = True, cmap = "coolwarm")
```

# Relationship between categorical columns

```python
# using Pivot table : to find relationship between categorical variables
# Heatmap: make the colorful table for showing pivot table

# categorical columns : sex, day, time, smoker

pivot = df.pivot_table(index = ["day", "time"], columns = ["sex", "smoker"], aggfunc = "size", observed=True, fill_value = 0)
pivot
```

```python
plt.figure(figsize = (3,2))
sns.heatmap(pivot, annot = True, cmap = "Reds")
```
