# match case

```python
# Match case is used to replace multiple if else condition
# Match case is more powerful than the Switch case statment
# Match case works with variable and data structurs. 
```

```python
# Example-1:
num = int(input("Enter a number"))

match num:
    case 10:
        print("You entered 10")
    case 20:
        print("You entered 20")
    case 30:
        print("You entered 30")
    case _:
        print("something else")
```

```python
Biryani = input("Enter Biryani name")

match Biryani:
    case "Paradise":
        print("Spicy & Tasty")
    case "Shahdab":
        print("Double masala")
    case "road side":
        print("Budget friendly")
    case _:
        print("PG Biryani")
```

```python
# Example-2: passing data structure to match case

data = [1,2,3,4]

match data:
    case [4,3,2,1]:
        print("Hello")
    case [1,2,3,4]:
        print("world")
```

```python
# Example-3: if condition with match case

val = int(input("Enter a number"))

match val:
    case x if x > 0:
        print("It is a positive numner")
    case y if y < 0:
        print("It is a Negative Number")
    case _:
        print("It is a Zero")
```

# Decorators

```python
def decorator_fun(func):   # func => cake()
    def wrapper_fun():  # wrapper function add the extra functionality
        print("Buy the ingredients")
        func()
        print("add cherry")    # Extra functionality
        print("add chocolate")
    return wrapper_fun

def cake():
    print("take the flour")
    print("add eggs and other ingredient")
    print("Put the cake in oven")

decorated_function = decorator_fun(cake)
decorated_function()
```

```python
# example-2 : use of @ to represent the decorators

def decorator_fun(func):   # func => cake()
    def wrapper_fun():  # wrapper function add the extra functionality
        print("Buy the ingredients")
        func()
        print("add cherry")    # Extra functionality
        print("add chocolate")
    return wrapper_fun

@decorator_fun
def cake():
    print("take the flour")
    print("add eggs and other ingredient")
    print("Put the cake in oven")

cake()
```

```python
# Multiple decorator for same function
def decorator_fun1(func):   # func => cake()
    def wrapper_fun():  # wrapper function add the extra functionality
        print("Buy the ingredients")
        func()
    return wrapper_fun

def decorator_fun2(func):   # func => cake()
    def wrapper_fun():  # wrapper function add the extra functionality
        func()
        print("add cherry")    # Extra functionality
        print("add chocolate")
    return wrapper_fun

@decorator_fun1
@decorator_fun2
def cake():
    print("take the flour")
    print("add eggs and other ingredient")
    print("Put the cake in oven")

cake()
```

#### Encapsulation

```python
# Def: Encapsulation refers to the binding of data (attributes)
#       and functions into one unit (class).

# Encapsultation restrict the direct access of the data.

# We can access the variables/data/attribute through object outside the class.
# We can access the variables/data/attribute through functions defined inside the class.

class Bike:
    No_tyre = 2   # class variable / Attribute / data

    def __init__(self, name, price):
        self.bike_name = name   # instance variable
        self.price = price

    def display(self):
        print(self.No_tyre,self.bike_name,self.price )

obj1 = Bike("RE", 1.5)

obj1.display()

print(obj1.No_tyre)
print(obj1.bike_name)
print(obj1.price)
```

### 1. Function overloading

#### using default arguments for function overloading

```python
# example-1
def fun(var1, var2 = None, var3 = None):
    if var2 is not None:
        print(var1, var2, var3)
    else:
        print(var1)

fun(10)
fun(10,20,30)
```

```python
# example-2 : default argument
def calculator(a = 0, b = 0, c = 0, d = 0):
    return a + b + c + d

print(calculator())
print(calculator(100))
print(calculator(100, 200))
print(calculator(100, 200, 300))
print(calculator(100, 200, 300, 400))
```

```python
## *args for function overloading
def calculator(*arg):
    return sum(arg)

print(calculator())
print(calculator(100))
print(calculator(100, 200))
print(calculator(100, 200, 300))
print(calculator(100, 200, 300, 400, 500, 6000, 7, 8, 9, 10, 11))
```

# Operator overloading

```python
# we will perform different operations on the user defined objects using operators.
```

```python
class Dosa:
    def __init__(self, variety, price):
        self.Variety = variety
        self.Price = price
    def __add__(self, d2_val):
        total_price = self.Price + d2_val.Price
        return total_price
    def __sub__(self, d2_val):
        total_price = self.Price - d2_val.Price
        return total_price

    def __mul__(self, d2_val):
        total_price = self.Price * d2_val.Price
        return total_price

    def __truediv__(self, d2_val):
        total_price = self.Price / d2_val.Price
        return total_price

    def __floordiv__(self, d2_val):
        total_price = self.Price // d2_val.Price
        return total_price

    def __mod__(self, d2_val):
        total_price = self.Price % d2_val.Price
        return total_price
d1 = Dosa("Masala", 60)
d2 = Dosa("IceCream", 110)

print(d1 + d2)
print(d1 - d2)
print(d1 * d2)  # __mul__
print(d2 / d1)  # __truediv__
print(d2 // d1) # __floordiv__
print(d1 % d2) # __mod__
# print(d1 ** d2) # __pow__
```

```python
# comparison operators overloading

class Dosa:
    def __init__(self, variety, price):
        self.Variety = variety
        self.Price = price
    def __eq__(self, d2_val):
        total_price = self.Price == d2_val.Price
        return total_price
    def __gt__(self, d2_val):
        total_price = self.Price > d2_val.Price
        return total_price

    def __lt__(self, d2_val):
        total_price = self.Price < d2_val.Price
        return total_price

    def __ge__(self, d2_val):
        total_price = self.Price >= d2_val.Price
        return total_price

    def __le__(self, d2_val):
        total_price = self.Price <= d2_val.Price
        return total_price

d1 = Dosa("Masala", 60)
d2 = Dosa("IceCream", 110)

print(d1 > d2) # __gt__()
print(d1 < d2) # __lt__()
print(d1 == d2) # __eq__()
print(d1 >= d2) # __ge__()
print(d1 <= d2) # __le__()
```

# Data Abstraction

```python
# Example-1: how to define abstract class, abstract method

from abc import ABC, abstractmethod

# abstract class
class Head_Mahindra(ABC):
    @abstractmethod
    def air_bag(self):  # abstract method
        pass

    @abstractmethod
    def LED_Screen(self):
        pass

    def insurance(self):  # concrete
        print("Please take Mahindra company insurance")

class hyd_mahindra(Head_Mahindra): # child class
    def air_bag(self):  # abstract method
        print("I will use Nylon to design the Air bag")

    def LED_Screen(self):
        print("Sony LED")

class noida_mahindra(Head_Mahindra):
    def air_bag(self):  # abstract method
        print("I will use Cotton to design the Air bag")

    def LED_Screen(self):
        print("Samsung LED")

obj_hyd = hyd_mahindra()   # create object of hyd child class
obj_noida = noida_mahindra()

obj_hyd.insurance()
# obj_head = Head_Mahindra()
```

```python
# Type of blocks in Exception Handling:
# 1. try : put that code where error can occur
# 2. except : handle the error 
# 3. else : put code which you want to execute when their is NO error
# 4. finally : execute either error or No error
```

```python
# Example-2: Exception as e

try:
    print(y)
except Exception as e:
    print(f"compiler error : {e}")
    print(type(e))

print("hello")
a,b = 10, 20
print(a+b)
```

```python
# Example-3: else block: execute if NO error

try:
    y = 20
    print(y)
except Exception as e:
    print(f"compiler error : {e}")
    print(type(e))
else:
    print("Hurry, we successfully executed code")
```

```python
# Example-4: finally block: execute if NO error + Error

try:
    y = 20
    print(z)
except Exception as e:
    print(f"compiler error : {e}")
    print(type(e))
else:
    print("Hurry, we successfully executed code")
finally:
    print("I will execute whatever happen")
```
