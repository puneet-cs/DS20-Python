# Creating Numpy array

```python
# 1D array
count_food_items = np.array([25, 95, 10, 12, 15, 20, 40.8, "hello"])
print(count_food_items)
print(type(count_food_items))
print(count_food_items.dtype)
```

```python
# 2D array
arr2 = np.array([["Lungi", "Kurta", "Puncha"],["Saree", "Half Saree", "Kurti"]])

print(arr2)
print(type(arr2))
print(arr2.dtype)
```

```python
arr3 = np.array([[1,2,3], [4,5,6]])

print(arr3)
print(arr3.ndim)
print(arr3.shape)
```

```python
# 3D array
arr4 = np.array([ [[1,2,3], 
                   [4,5,6]], 
                 
                  [[1,2,3], 
                   [4,5,6]] ])

print(arr4)
print(arr4.ndim)
print(arr4.shape)
```

```python
# 4D
arr4 = np.array([ [ [[1,2,3], 
                   [4,5,6]], 
                 
                  [[1,2,3], 
                   [4,5,6]] ],

                [ [[1,2,3], 
                   [4,5,6]], 
                 
                  [[1,2,3], 
                   [4,5,6]] ] ]
               
               )

print(arr4)
print(arr4.ndim)
print(arr4.shape)
```

```python
# 5D
arr4 = np.array([[ [ [[1,2,3,8,9], 
                   [4,5,6,6,7]], 
                 
                  [[1,2,3,7,8], 
                   [4,5,6,6,7]] ],

                [ [[1,2,3,5,6], 
                   [4,5,6,5,6]], 
                 
                  [[1,2,3,3,4], 
                   [4,5,6,1,2]] ] ],

                [ [ [[1,2,3,8,9], 
                   [4,5,6,7,8]], 
                 
                  [[1,2,3,4,5], 
                   [4,5,6,3,4]] ],

                [ [[1,2,3,4,5], 
                   [4,5,6,4,5]], 
                 
                  [[1,2,3,3,4], 
                   [4,5,6,7,8]] ] ]]
               
               )

print(arr4)
print(arr4.ndim)
print(arr4.shape)
```

# Special type of array

```python
# 1D array of zeros
arr6 = np.zeros(5)
print(arr6)
```

```python
# 2D zeros array
arr7 = np.zeros((2,3,3))
print(arr7)
```

```python
arr_ones = np.ones(5)
print(arr_ones)
```

```python
arr_ones = np.ones((2,5))
print(arr_ones)
```

```python
arr9 = np.full(4, 1000)
print(arr9)
```

```python
arr9 = np.full((2,4), 50)
print(arr9)
```
