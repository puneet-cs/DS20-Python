# Performance matrics

# Confusion matrix

```python
# Libraries
from sklearn.metrics import confusion_matrix
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.datasets import load_breast_cancer
```

```python
# load the inbuild dataset
cancer_obj = load_breast_cancer()
dir(cancer_obj)
```

```python
print(cancer_obj.DESCR)
```

```python
# store the input features data into x and target data into y  varaibles
x = cancer_obj.data
y = cancer_obj.target
```

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size = 0.2, random_state = 42)
```

```python
# scaling of data
scaler = StandardScaler()
x_train_scaled = scaler.fit_transform(x_train)
x_test_scaled = scaler.transform(x_test)
```

```python
# import different ML algorithm
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.neighbors import KNeighborsClassifier
from sklearn.naive_bayes import GaussianNB
from sklearn.svm import SVC
```

```python
# model object creation
logreg = LogisticRegression()
dtree = DecisionTreeClassifier()
knn = KNeighborsClassifier()
nb = GaussianNB()
svm = SVC()
```

```python
# training of all the ML algorithm
logreg.fit(x_train_scaled, y_train)
dtree.fit(x_train_scaled, y_train)
knn.fit(x_train_scaled, y_train)
nb.fit(x_train_scaled, y_train)
svm.fit(x_train_scaled, y_train)
```

```python
# score 
print(logreg.score(x_test_scaled, y_test))
print(dtree.score(x_test_scaled, y_test))
print(knn.score(x_test_scaled, y_test))
print(nb.score(x_test_scaled, y_test))
print(svm.score(x_test_scaled, y_test))
```

```python
# making prediction
y_pred_logreg = logreg.predict(x_test_scaled)
y_pred_dtree = dtree.predict(x_test_scaled)
y_pred_knn = knn.predict(x_test_scaled)
y_pred_nb = nb.predict(x_test_scaled)
y_pred_svm = svm.predict(x_test_scaled)
```

```python
# y_test is the "actual" target value
import seaborn as sns
import matplotlib.pyplot as plt
# construct the confusion matrix
plt.figure(figsize = (3,2))
cm_logreg = confusion_matrix(y_pred_logreg, y_test)
sns.heatmap(cm_logreg, annot = True)
plt.ylabel("Predicted Values")
plt.xlabel("Actual values")
```

```python
plt.figure(figsize = (3,2))
cm_dtree = confusion_matrix(y_test, y_pred_dtree)
sns.heatmap(cm_dtree, annot = True)
plt.xlabel("Predicted Values")
plt.ylabel("Actual values")
```

```python
plt.figure(figsize = (3,2))
cm_knn = confusion_matrix(y_test, y_pred_knn)
sns.heatmap(cm_knn, annot = True)
plt.xlabel("Predicted Values")
plt.ylabel("Actual values")
```

```python
plt.figure(figsize = (3,2))
cm_nb = confusion_matrix(y_test, y_pred_nb)
sns.heatmap(cm_nb, annot = True)
plt.xlabel("Predicted Values")
plt.ylabel("Actual values")
```

```python
plt.figure(figsize = (3,2))
cm_svm = confusion_matrix(y_test, y_pred_svm)
sns.heatmap(cm_svm, annot = True)
plt.xlabel("Predicted Values")
plt.ylabel("Actual values")
```
